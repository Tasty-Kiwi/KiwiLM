## 4. Prepare or restore the 550k-story dataset

Model C has 6,050,816 parameters, so the 20:1 target is 121,016,320 training tokens. The current tokenizer averages about 222.2 tokens/story; 550k stories provides a small margin over the estimated 544,631-story target.

The first run streams TinyStories and trains a new 8k BPE tokenizer. The prepared artifacts are then copied to Drive. Later Colab sessions restore the exact prepared files from Drive instead of rebuilding them.

```python
%%bash
set -euo pipefail
cd /content/KiwiLM
DRIVE_DATA=/content/drive/MyDrive/KiwiLM/data/tinystories-550k
LOCAL_DATA=/content/KiwiLM/data/tinystories-550k
mkdir -p /content/drive/MyDrive/KiwiLM/data /content/KiwiLM/data

if [[ -f "$DRIVE_DATA/metadata.json" ]]; then
  echo "Restoring prepared data from Google Drive..."
  cp -a "$DRIVE_DATA" /content/KiwiLM/data/
elif [[ ! -f "$LOCAL_DATA/metadata.json" ]]; then
  uv run kiwilm prepare \
    --output-dir "$LOCAL_DATA" \
    --train-limit 550000 \
    --validation-limit 10000 \
    --vocab-size 8192
  echo "Saving prepared data to Google Drive..."
  cp -a "$LOCAL_DATA" /content/drive/MyDrive/KiwiLM/data/
fi

du -sh "$LOCAL_DATA"
test -f "$LOCAL_DATA/metadata.json"
```

## 5. Start a new training run

Batch 64 × context 256 × 7,400 steps processes 121,241,600 tokens, or 20.04 tokens per parameter. Batch 64 improves GPU utilization while fitting comfortably for this 6.05M-parameter model.

Run this cell only for a new run. It refuses to overwrite an existing Drive checkpoint. Checkpoints are saved every 500 steps, so a disconnect loses at most 500 steps.

```python
%%bash
set -euo pipefail
cd /content/KiwiLM
RUN_DIR=/content/drive/MyDrive/KiwiLM/runs/model-c-chinchilla
mkdir -p "$RUN_DIR"

if [[ -f "$RUN_DIR/latest.pt" ]]; then
  echo "An existing checkpoint was found. Run the resume cell instead."
  exit 1
fi

TORCH_ALLOW_TF32_CUBLAS_OVERRIDE=1 uv run kiwilm train \
  --architecture cnn_dual_attention \
  --data-dir /content/KiwiLM/data/tinystories-550k \
  --output-dir "$RUN_DIR" \
  --device cuda \
  --batch-size 64 \
  --max-steps 7400 \
  --warmup-steps 370 \
  --eval-interval 500 \
  --eval-batches 50 \
  --checkpoint-interval 500 \
  --log-interval 10 \
  --seed 42
```

## 6. Resume after a disconnect

In a fresh Colab session, rerun the GPU, Drive, setup, and dataset-restore cells first. Then run this cell instead of the new-training cell. The CLI restores optimizer, scheduler, random-generator, and batch-sampler state.

```python
%%bash
set -euo pipefail
cd /content/KiwiLM
RUN_DIR=/content/drive/MyDrive/KiwiLM/runs/model-c-chinchilla
test -f "$RUN_DIR/latest.pt"

TORCH_ALLOW_TF32_CUBLAS_OVERRIDE=1 uv run kiwilm train \
  --architecture cnn_dual_attention \
  --data-dir /content/KiwiLM/data/tinystories-550k \
  --output-dir "$RUN_DIR" \
  --resume "$RUN_DIR/latest.pt" \
  --device cuda \
  --batch-size 64 \
  --max-steps 7400 \
  --warmup-steps 370 \
  --eval-interval 500 \
  --eval-batches 50 \
  --checkpoint-interval 500 \
  --log-interval 10 \
  --seed 42
```

## 7. Evaluate the best checkpoint

```python
!uv run kiwilm evaluate --data-dir /content/KiwiLM/data/tinystories-550k --checkpoint /content/drive/MyDrive/KiwiLM/runs/model-c-chinchilla/best.pt --device cuda --batch-size 64 --batches 200 --seed 42
```
